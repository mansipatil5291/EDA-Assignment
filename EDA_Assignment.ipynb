Question 1 : What is the difference between AI, ML, DL, and Data Science? Provide a brief explanation of each. (Hint: Compare their scope, techniques, and applications for each.)

-> Think of these as layers of a Russian nesting doll, with Data Science acting as the umbrella that uses them all. Here is the breakdown:

1. Artificial Intelligence (AI):

* Scope: The broadest category. It aims to create machines capable of mimicking human intelligence and decision-making.

* Techniques: Includes everything from simple "if-then" logic (Rule-based systems) to advanced neural networks.

* Applications: Virtual assistants (Siri/Alexa), self-driving cars, and game-playing bots (Chess/Go).

2. Machine Learning (ML):

* Scope: A subset of AI. It focuses on the idea that systems can learn from data and improve from experience without being explicitly programmed for every task.

* Techniques: Statistical algorithms like Linear Regression, Decision Trees, and Support Vector Machines.

* Applications: Email spam filters, product recommendations (Netflix/Amazon), and credit scoring.

3. Deep Learning (DL):

* Scope: A subset of ML inspired by the structure of the human brain. It handles massive amounts of unstructured data (like images and sound).

* Techniques: Multi-layered Neural Networks (Artificial Neural Networks, CNNs, RNNs).

* Applications: Facial recognition, real-time language translation, and Generative AI (ChatGPT/DALL-E).

4. Data Science:

* Scope: An interdisciplinary field that overlaps with all the above. It’s not just about building models; it’s about extracting actionable insights from data to solve business problems.

* Techniques: Combines ML and DL with data visualization, statistics, data cleaning, and "big data" processing.

* Applications: Identifying market trends, optimizing supply chains, and healthcare diagnostics.

Question 2: Explain overfitting and underfitting in ML. How can you detect and prevent them? Hint: Discuss bias-variance tradeoff, cross-validation, and regularization techniques.

-> In Machine Learning, building a model is like a student studying for an exam: you want them to learn the concepts, not just memorize the practice test.

1. Overfitting (High Variance)

* The Problem: The model learns the training data too well, including the noise and random fluctuations. It performs great on seen data but fails on new, unseen data.

* Detection: High accuracy on the training set but low accuracy on the validation/test set.

* Prevention: 1) Regularization (L1/L2): Adds a penalty for overly complex models to keep weights small. 2) More Data: Helps the model ignore noise and find the true signal. 3) Pruning: In decision trees, cutting back branches that provide little predictive power. 4) Dropout: In deep learning, randomly "turning off" neurons during training.

2. Underfitting (High Bias)

* The Problem: The model is too simple to capture the underlying patterns in the data. It’s like trying to fit a straight line to a curved trend.

* Detection: Low accuracy on both the training set and the validation set.

* Prevention: 1) Increase Complexity: Use a more powerful algorithm (e.g., switching from Linear Regression to a Neural Network). 2) Feature Engineering: Add more relevant input features or perform polynomial transformations. 3) Reduce Regularization: If the "penalty" for complexity is too high, the model stays too simple.

3. Bias-Variance Tradeoff:

* Bias is the error from overly simplistic assumptions (Underfitting).

* Variance is the error from over-sensitivity to small fluctuations (Overfitting).

* The goal is to find the "sweet spot" where both are minimized.

4. Cross-Validation:

* Instead of one test, you split your data into k folders and rotate which one is the "test" set. This ensures the model's performance is consistent across different data slices and isn't just "getting lucky."

Question 3:How would you handle missing values in a dataset? Explain at least three methods with examples. Hint: Consider deletion, mean/median imputation, and predictive modeling.

-> Handling missing values is a critical step because most ML algorithms can't process "NaN" or "Null" entries. The best approach depends on the amount of data you have and the nature of the missingness.

1. Deletion (Dropping)

* How it works: You remove the entire row or column containing the missing value.

* When to use: Use this when the missing data is very small (less than 5%) or when a column is missing so much data (e.g., >60%) that it’s useless.

* Example: In a survey of 1,000 people, if only 2 people didn’t provide their "Age," you can safely delete those two rows without losing significant insight.

2. Mean/Median/Mode Imputation

* How it works: You fill the gaps with a statistical measure of the existing data in that column. 1) Mean: Good for numerical data without outliers (e.g., Average Height). 2) Median: Better for numerical data with outliers (e.g., Household Income). 3) Mode: Used for categorical data (e.g., Most common "City").

* Example: If a dataset of house prices is missing values for "Number of Bedrooms," you could fill them with the median (e.g., 3) to represent a typical house.

3. Predictive Imputation (Modeling)

* How it works: You treat the missing value as a "target" and use other available features to predict it using an algorithm like K-Nearest Neighbors (KNN) or Random Forest.

* When to use: Use this when there is a strong correlation between features, and you want a more sophisticated, accurate guess than just a simple average.

* Example: If "Weight" is missing, a model can predict it based on the person's "Height," "Gender," and "Age" rather than just giving everyone the same average weight.

Question 4:What is an imbalanced dataset? Describe two techniques to handle it (theoretical + practical). Hint: Discuss SMOTE, Random Under/Oversampling, and class weights in models.

-> An imbalanced dataset occurs when the classes you are trying to predict are not represented equally. One class (the majority class) significantly outnumbers the other (the minority class). This is a major problem because ML models are "lazy"—they will often just predict the majority class every time to achieve high accuracy, while completely failing to identify the rare cases you actually care about (like fraud or rare diseases).

Here are two primary ways to fix it:

1. Resampling Techniques (Data-Level): These methods physically change the size of your dataset to balance the proportions.

* Random Undersampling: You delete examples from the majority class until it matches the minority. Risk: You might throw away useful information.

* Random Oversampling: You duplicate examples from the minority class. Risk: This often leads to overfitting because the model just memorizes specific rows.

* SMOTE (Synthetic Minority Over-sampling Technique): This is the "smarter" version of oversampling. Instead of just duplicating rows, it looks at the minority class and creates new, synthetic examples by interpolating between existing ones. Practical Example: If you have 10 fraudulent transactions and 1,000 legitimate ones, SMOTE creates "fake" fraud cases that look similar to the real ones to give the model more to learn from.

2. Algorithmic Adjustments (Model-Level): Instead of changing the data, you change how the model learns.

* Class Weights: You tell the model that the minority class is "more important." In the loss function, a mistake on a minority class instance is penalized more heavily than a mistake on a majority class instance.

Practical Example: In a Logistic Regression or Random Forest, you can set a parameter like class_weight='balanced'. This mathematically "boosts" the signal of the rare class so the model pays more attention to it.

The "Golden Rule" of Imbalanced Data:

Never use Accuracy as your metric! A model that predicts "No Fraud" 100% of the time on a 99% legitimate dataset is 99% accurate, but it’s completely useless. Use Precision, Recall, or the F1-Score instead.

Question 5: Why is feature scaling important in ML? Compare Min-Max scaling and Standardization. Hint: Explain impact on distance-based algorithms (e.g., KNN, SVM) and gradient descent.

-> Feature scaling is essential because most ML algorithms are sensitive to the magnitude of your data. If one feature ranges from 0 to 1 (like "Probability") and another from 0 to 1,000,000 (like "House Price"), the larger numbers will dominate the model's math, even if they aren't more important.

1. Why it matters (The Impact)

* Distance-Based Algorithms (KNN, SVM, K-Means): These rely on calculating the Euclidean distance between points. Without scaling, the "House Price" would make the "Probability" feature mathematically invisible.

* Gradient Descent (Linear/Logistic Regression, Neural Networks): Scaling helps the optimization process. If features have wildly different scales, the gradient descent "path" to the minimum error becomes jagged and slow. Scaling makes the path smooth and fast.

* Note: Tree-based models (Random Forest, XGBoost) are generally invariant to scaling, as they split data based on rank, not distance.

2. Min-Max Scaling vs. Standardization

1)Min-Max Scaling (Normalization)

-Formula: (x-min)/(max-min)

-Output Range: Strictly 0 to 1 (usually).

-Outliers: Very sensitive; one outlier can squash all other data into a tiny range.

-Best Use Case: When you know the bounds (e.g., Image pixels 0–255) or use algorithms like Neural Networks.

2)Standardization (Z-Score)

-Formula: (x-mean)/standard_deviation

-Output Range: Not bounded; centered at 0 with a std dev of 1.

-Outliers: More robust; it handles outliers better by centering data.

-Best Use Case: When data follows a Gaussian (Normal) distribution or for PCA and SVMs.

3. Practical Tip:

If you aren't sure which to pick, Standardization is generally the "safer" default for most general ML tasks because it is less affected by extreme values.

Question 6: Compare Label Encoding and One-Hot Encoding. When would you prefer one over the other? Hint: Consider categorical variables with ordinal vs. nominal relationships.

-> Machine Learning models only speak the language of numbers, so we have to translate text categories into a numerical format. The "right" translation depends on whether the categories have a natural order.

1. Label Encoding (Numerical Mapping):

* How it works: Each unique category is assigned a unique integer (e.g., Apple = 0, Banana = 1, Cherry = 2).

* Best for: Ordinal Data, where the categories have a clear rank or hierarchy.

* Example: "Education Level" (High School = 1, Bachelor's = 2, Master's = 3). The model needs to understand that a Master's is "greater" than High School.

* The Risk: If used on non-ordered data (like "Colors"), the model might mistakenly think "Cherry (2)" is somehow "greater than" or "better than" "Apple (0)."

2. One-Hot Encoding (Binary Columns):

* How it works: It creates a new "dummy" column for every unique category. Each row gets a 1 in the column it belongs to and a 0 in all others.

* Best for: Nominal Data, where there is no inherent order or relationship between categories.

* Example: "City" (London, Paris, New York). A 1 in the "London" column doesn't imply it is "more" than Paris; they are just distinct.


* The Risk: The "Curse of Dimensionality." If you have a column with 1,000 unique cities, One-Hot Encoding will add 1,000 new columns to your dataset, making it massive and slow to train.

3. When to choose which?

* Ordinal Data (Small, Medium, Large) - Label Encoding - Preserves the mathematical relationship/rank.

* Nominal Data (Red, Blue, Green) - One-Hot Encoding - Prevents the model from assuming a fake rank.

* High Cardinality (>100 unique categories) - Binary/Target Encoding - One-Hot would create too many columns.

Answer of Question 7: Google Play Store Dataset.

Analysis of the Google Play Store dataset reveals a significant relationship between app categories and user ratings, with average ratings across most categories typically clustering between 4.0 and 4.5

1. Highest and Lowest Rated Categories:

* Highest Rated Categories: Categories such as EDUCATION, HEALTH_AND_FITNESS, and BOOKS_AND_REFERENCE frequently exhibit the highest average ratings.

* Lowest Rated Categories: Categories like DATING, TOOLS, and MAPS_AND_NAVIGATION often show lower average ratings compared to others.

2. Possible Reasons for Rating Variations:

* Specific Utility & Value: High-rated categories like Education or Health often provide highly specific, focused value that meets clear user expectations.

* User Frustration & Technical Issues: Lower-rated categories often face higher scrutiny. For example, DATING apps may receive lower ratings due to user disappointment with matching algorithms or social interactions, while TOOLS often suffer from intrusive ads or technical bugs.

* App Quality vs. Price: Evidence suggests that paid apps tend to have slightly higher median sentiment and quality perception compared to free apps, which may encounter more negative feedback due to lower entry barriers and higher ad frequency.


```
# Answer of Question 7: Google Play Store Dataset.

import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset (Assuming the standard googleplaystore.csv structure)
df = pd.read_csv('googleplaystore.csv')

# 1. Data Cleaning
# Remove row with invalid 'Rating' (e.g., 19) and handle NaN values
df = df[df['Rating'] <= 5]
df['Rating'] = df['Rating'].fillna(df['Rating'].mean())

# 2. Relationship Analysis: Average Rating per Category
category_ratings = df.groupby('Category')['Rating'].mean().sort_values(ascending=False)

# 3. Visualization
plt.figure(figsize=(12, 6))
category_ratings.plot(kind='bar', color='skyblue')
plt.title('Average App Rating by Category')
plt.ylabel('Average Rating')
plt.xlabel('Category')
plt.ylim(3.5, 5)  # Zoom in to see differences
plt.xticks(rotation=90)
plt.show()

# Output the highest and lowest
print(f"Highest Rated Category: {category_ratings.idxmax()} ({category_ratings.max():.2f})")
print(f"Lowest Rated Category: {category_ratings.idxmin()} ({category_ratings.min():.2f})")
```

Answer of Question 8: Titanic Dataset.

-> Based on the analysis of the Titanic dataset, the survival rates were heavily influenced by passenger class and age groups.

a) Survival Rates by Passenger Class (Pclass):

The 1st Class had the highest survival rate, followed by 2nd and then 3rd Class.

* 1st Class: ~63%

* 2nd Class: ~47%

* 3rd Class: ~24%

Reasoning: First-class passengers likely had the highest survival rate due to their socio-economic privilege, which provided them with cabins located closer to the boat deck and prioritized access to lifeboats.

b) Impact of Age on Survival:

Children (Age < 18) had a significantly higher chance of survival compared to adults.

* Children Survival Rate: ~54%

* Adults Survival Rate: ~38%

Reasoning: This higher survival rate for children is largely attributed to the "women and children first" maritime protocol followed during the evacuation, ensuring that younger and more vulnerable passengers were prioritized for the limited space on lifeboats.

```
# Answer of Question 8: Titanic Dataset.

import pandas as pd

# Load the Titanic dataset
url = "https://githubusercontent.com"
df = pd.read_csv(url)

# --- Part A: Survival by Passenger Class ---
pclass_survival = df.groupby('Pclass')['Survived'].mean() * 100

print("Survival Rate by Class:")
print(pclass_survival)

# --- Part B: Survival by Age Group ---
# Create Age groups, dropping rows with missing Age values for accuracy
df_age = df.dropna(subset=['Age']).copy()
df_age['AgeGroup'] = df_age['Age'].apply(lambda x: 'Child' if x < 18 else 'Adult')

age_survival = df_age.groupby('AgeGroup')['Survived'].mean() * 100

print("\nSurvival Rate by Age Group:")
print(age_survival)
```

Answer of Question 9: Flight Price Prediction Dataset.

-> Based on the Flight Price Prediction dataset, which contains data sourced from EaseMyTrip, the analysis of flight prices reveals significant trends regarding booking time and airline positioning.

a) Impact of Days Left Until Departure:

Flight prices are inversely related to the number of days left until departure.

* Exponential Price Surges: Prices remain relatively stable when booked more than 20 days in advance. However, as the departure date approaches—specifically within the 1–2 day window—prices experience an exponential surge due to high demand and limited last-minute seat availability.

* Best Booking Window: To secure the most affordable fares, it is recommended to book between 20 and 50 days before departure. Booking at least 3 weeks early typically avoids the steepest price escalations.

b) Airline Comparison (Delhi-Mumbai Route):

On the popular Delhi-Mumbai route, airlines are generally split into two tiers: economical and premium.

* Economical Airlines: IndiGo, SpiceJet, and AirAsia are consistently the cheapest options. These airlines operate on a Low-Cost Carrier (LCC) model, focusing on high-frequency, point-to-point travel with fewer complimentary amenities.

* Premium Airlines: Vistara and Air India consistently command higher prices.

1)Vistara is often the most expensive as it offers a full-service experience, including premium economy and business class options.

2)Air India maintains higher average fares due to its full-service model, including complimentary meals and higher baggage allowances.

```
# Answer of Question 9: Flight Price Prediction Dataset.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset (Assumes local path to the MasteriNeuron repo file)
df = pd.read_csv('Cleaned_Dataset.csv')

# a) How flight prices vary with days left
plt.figure(figsize=(10,6))
sns.lineplot(data=df, x='days_left', y='price')
plt.title('Flight Price vs Days Left Until Departure')
plt.xlabel('Days Left')
plt.ylabel('Average Price')
plt.show()

# b) Compare prices across airlines for Delhi-Mumbai
delhi_mumbai = df[(df['source_city'] == 'Delhi') & (df['destination_city'] == 'Mumbai')]
plt.figure(figsize=(10,6))
sns.boxplot(data=delhi_mumbai, x='airline', y='price')
plt.title('Price Distribution by Airline (Delhi to Mumbai)')
plt.show()

# Output summary statistics
print(delhi_mumbai.groupby('airline')['price'].mean().sort_values())
```

-> Based on the IBM HR Analytics Employee Attrition & Performance dataset, attrition is driven by a mix of workload, compensation, and satisfaction metrics.

a) Key Drivers of Employee Attrition:

The following factors show the strongest correlation with employees choosing to leave:

* Overtime: This is often the single strongest predictor of attrition. Employees working overtime are nearly three times more likely to leave compared to those who do not.

* Monthly Income: There is a strong negative correlation between salary and attrition. Employees earning lower monthly incomes (specifically under $5,000) represent the bulk of departures.

* Job & Environment Satisfaction: Low ratings in these areas are critical drivers. Attrition rates significantly drop as job satisfaction levels increase from 1 (Low) to 4 (Very High).

* Tenure & Age: Younger employees (ages 20–30) and those in their first 2–3 years at the company have the highest attrition rates.

b) Relationship Between Projects and Attrition:

Employees with a very high number of projects are significantly more likely to leave, typically due to burnout.

* Extreme Workloads: In variations of this dataset, 100% of employees assigned to 7 projects eventually left the company.

* Underutilization Risk: Conversely, those with very few projects (e.g., 2) also show higher turnover, suggesting that both overworked and underworked employees are at risk.

* Optimal Range: Employees assigned to 3 to 5 projects are the most likely to stay with the organization.

```
# Answer of Question 10: HR Analytics Dataset.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')

# a) Visualizing Key Drivers: Overtime and Monthly Income
plt.figure(figsize=(12, 5))

# Plot 1: Overtime vs Attrition
plt.subplot(1, 2, 1)
sns.countplot(data=df, x='OverTime', hue='Attrition')
plt.title('Attrition by Overtime Status')

# Plot 2: Monthly Income vs Attrition
plt.subplot(1, 2, 2)
sns.boxplot(data=df, x='Attrition', y='MonthlyIncome')
plt.title('Monthly Income Distribution by Attrition')
plt.show()

# b) Projects/Workload Proxy (using JobLevel or YearsAtCompany if project_count is unavailable)
# Note: In the standard IBM dataset, 'JobLevel' often correlates with project complexity.
plt.figure(figsize=(8, 5))
sns.pointplot(data=df, x='JobLevel', y='Attrition', join=False)
plt.title('Attrition Rate by Job Level')
plt.show()
```
